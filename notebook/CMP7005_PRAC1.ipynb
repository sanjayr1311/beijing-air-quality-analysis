## 3.2 Data Preprocessing

### 3.2.1 Create a Copy of the Dataset

```python
df = merged_df.copy()

print(f"Dataset Shape: {df.shape}")
```

### 3.2.2 Handle Missing Values

```python
# Numerical columns
numerical_columns = df.select_dtypes(include=["int64", "float64"]).columns

# Exclude identifier columns
numerical_columns = numerical_columns.drop(["No", "year", "month", "day", "hour"])

for column in numerical_columns:
    df[column].fillna(df[column].median(), inplace=True)

# Categorical column
df["wd"].fillna(df["wd"].mode()[0], inplace=True)

# Verify Missing Values

missing_after = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values
})

display(missing_after)
```

### 3.2.3 Remove Duplicate Records

```python
# no Duplicate records were forund.
```

### 3.2.4 Datetime Feature Extraction

```python
df["Day_of_Week"] = df["datetime"].dt.day_name()

df["Weekday"] = df["datetime"].dt.weekday

df["Quarter"] = df["datetime"].dt.quarter

```

### 3.2.5 Create Season Feature

```python
def get_season(month):

    if month in [12,1,2]:
        return "Winter"

    elif month in [3,4,5]:
        return "Spring"

    elif month in [6,7,8]:
        return "Summer"

    else:
        return "Autumn"

df["Season"] = df["month"].apply(get_season)

print(df["Season"].value_counts())
```

### 3.2.6 Create AQI Category

```python
def aqi_category(pm25):

    if pm25 <= 12:
        return "Good"

    elif pm25 <= 35.4:
        return "Moderate"

    elif pm25 <= 55.4:
        return "Unhealthy for Sensitive Groups"

    elif pm25 <= 150.4:
        return "Unhealthy"

    elif pm25 <= 250.4:
        return "Very Unhealthy"

    else:
        return "Hazardous"

df["AQI_Category"] = df["PM2.5"].apply(aqi_category)

display(df["AQI_Category"].value_counts())
```

### 3.2.7 Encode Categorical Variables

```python
from sklearn.preprocessing import LabelEncoder

le_station = LabelEncoder()
le_wd = LabelEncoder()
le_season = LabelEncoder()
le_aqi = LabelEncoder()

df["station_encoded"] = le_station.fit_transform(df["station"])

df["wd_encoded"] = le_wd.fit_transform(df["wd"])

df["season_encoded"] = le_season.fit_transform(df["Season"])

df["aqi_encoded"] = le_aqi.fit_transform(df["AQI_Category"])

```

### 3.2.8 Outlier Detection

```python
pollutants = ["PM2.5","PM10","SO2","NO2","CO","O3"]

plt.figure(figsize=(14,8))

for i, column in enumerate(pollutants,1):

    plt.subplot(2,3,i)

    sns.boxplot(y=df[column])

    plt.title(column)

plt.tight_layout()

plt.show()
```

# 4. Task 3: Model Building

## 4.1 Define the Prediction Problem

```python

print("Prediction Problem")
print()

print("Target Variable : PM2.5")
print("Problem Type    : Regression")
print("Goal            : Predict PM2.5 concentration using meteorological and pollutant variables.")
```

## 4.2 Select Features and Target Variable

```python
features = [
    "PM10",
    "SO2",
    "NO2",
    "CO",
    "O3",
    "TEMP",
    "PRES",
    "DEWP",
    "RAIN",
    "WSPM",
    "station_encoded",
    "season_encoded",
    "Weekday",
    "hour"
]

X = df[features]

y = df["PM2.5"]

print("Feature Matrix Shape :", X.shape)
print("Target Shape :", y.shape)
```

## 4.3 Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Set :", X_train.shape)
print("Testing Set  :", X_test.shape)
```

## 4.4 Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

## 4.5 Build Machine Learning Models

```python
from sklearn.linear_model import LinearRegression

from sklearn.ensemble import RandomForestRegressor

from sklearn.ensemble import GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

### 4.5.1 Linear Regression

```python

lr = LinearRegression()

lr.fit(X_train_scaled, y_train)

lr_pred = lr.predict(X_test_scaled)

lr_mae = mean_absolute_error(y_test, lr_pred)

lr_rmse = mean_squared_error(y_test, lr_pred)**0.5

lr_r2 = r2_score(y_test, lr_pred)

print("Linear Regression Results")

print("MAE :", round(lr_mae,2))

print("RMSE :", round(lr_rmse,2))

print("R² :", round(lr_r2,4))
```

### 4.5.2 Random Forest Regressor

```python

rf = RandomForestRegressor(
    random_state=42,
    n_estimators=100
)

rf.fit(X_train, y_train)

rf_pred = rf.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_pred)

rf_rmse = mean_squared_error(y_test, rf_pred)**0.5

rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Results")

print("MAE :", round(rf_mae,2))

print("RMSE :", round(rf_rmse,2))

print("R² :", round(rf_r2,4))
```

### 4.5.3 Gradient Boosting Regressor

```python

gbr = GradientBoostingRegressor(
    random_state=42
)

gbr.fit(X_train, y_train)

gbr_pred = gbr.predict(X_test)

gbr_mae = mean_absolute_error(y_test, gbr_pred)

gbr_rmse = mean_squared_error(y_test, gbr_pred)**0.5

gbr_r2 = r2_score(y_test, gbr_pred)

print("Gradient Boosting Results")

print("MAE :", round(gbr_mae,2))

print("RMSE :", round(gbr_rmse,2))

print("R² :", round(gbr_r2,4))
```

## 4.6 Hyperparameter Optimization

```python
from sklearn.model_selection import GridSearchCV

parameters = {
    "n_estimators":[100,150],
    "max_depth":[10,20],
    "min_samples_split":[2,5]
}

grid = GridSearchCV(
    RandomForestRegressor(random_state=42),
    parameters,
    cv=3,
    scoring="r2",
    n_jobs=-1
)

grid.fit(X_train, y_train)

best_rf = grid.best_estimator_

print("Best Parameters")

print(grid.best_params_)
```

```python
# Evaluation
best_pred = best_rf.predict(X_test)

best_mae = mean_absolute_error(y_test,best_pred)

best_rmse = mean_squared_error(y_test,best_pred)**0.5

best_r2 = r2_score(y_test,best_pred)

print("Optimized Random Forest")

print("MAE :",round(best_mae,2))

print("RMSE :",round(best_rmse,2))

print("R² :",round(best_r2,4))
```

## 4.7 Model Comparison

```python
comparison = pd.DataFrame({

    "Model":[
        "Linear Regression",
        "Random Forest",
        "Gradient Boosting",
        "Optimized Random Forest"
    ],

    "MAE":[
        lr_mae,
        rf_mae,
        gbr_mae,
        best_mae
    ],

    "RMSE":[
        lr_rmse,
        rf_rmse,
        gbr_rmse,
        best_rmse
    ],

    "R2 Score":[
        lr_r2,
        rf_r2,
        gbr_r2,
        best_r2
    ]

})

comparison = comparison.round(4)

display(comparison)
```

## 4.8 Visual Comparison

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=comparison,
    x="Model",
    y="R2 Score"
)

plt.title("Model Comparison Based on R² Score")

plt.xticks(rotation=15)

plt.show()
```
